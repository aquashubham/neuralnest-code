## Why this matters

Have you ever tried to combine data from two tables, only to find half your rows suddenly missing? It happens to every developer, especially when you are just starting. In this video, we will use a college fest example to demystify SQL JOINs, so you can control exactly how your data merges.

## The College Fest Problem

- Students table: student ID, name.
- Events table: event ID, student ID, activity.
- Goal: list every student and their event.

Imagine we are organizing a college festival. We have a list of students and a list of registered events. Each event is linked to a student by their ID. The goal is simple: create a master report that shows every student alongside the activity they signed up for.

## Our Data Source

**Students**

| id | name |
|---|---|
| 1 | Alice |
| 2 | Bob |
| 3 | Charlie |

**Events**

| student_id | activity |
|---|---|
| 1 | Hackathon |
| 2 | Debate |

Look at these tables. Alice and Bob have events, but Charlie has not signed up for anything yet. If we just join them blindly, what happens to Charlie? We will see that in a moment.

## INNER JOIN Intuition

Two sets: **Students** and **Events**. Highlighted region: **inner**.

An INNER JOIN is like an exclusive club. It only returns rows where there is a match in both tables. Think of it as the intersection of two sets. If a student is not in the events table, they simply vanish from your results.

## Writing an INNER JOIN

Here is the SQL. We select the name and activity, joining on the ID column. Notice that Charlie is gone. This is exactly how I once lost important data when I first started—I assumed an INNER JOIN would give me everyone, but it only gives me those with a record in both places.

```sql
SELECT s.name, e.activity
FROM students s
JOIN events e ON s.id = e.student_id;
-- returns 2 rows: Alice, Bob
```

## What the Query Returns

**INNER JOIN result**

| name | activity |
|---|---|
| Alice | Hackathon |
| Bob | Debate |

Here is what the query returns. Alice and Bob are in the result, each with their event. But Charlie is nowhere to be found. If your boss asked for a list of all students, this result would be incomplete, and nothing would warn you.

## The LEFT JOIN Solution

Two sets: **Students** and **Events**. Highlighted region: **left**.

To include everyone, we use a LEFT JOIN. This tells SQL: give me everything from the left table—the students—even if there is no match on the right. When there is no match, SQL fills the empty space with a special value called NULL.

## Using LEFT JOIN

The syntax is almost identical, just add the word LEFT. Now, the result includes Charlie. The activity column for Charlie shows NULL, confirming that he exists in our system but has no entry in the events table.

```sql
SELECT s.name, e.activity
FROM students s
LEFT JOIN events e ON s.id = e.student_id;
# -> [('Alice', 'Hackathon'),
#     ('Bob', 'Debate'),
#     ('Charlie', NULL)]
```

## Key Takeaways

- INNER JOIN returns only matching records.
- LEFT JOIN keeps all rows from the primary table.
- NULL represents missing data for non-matching rows.
- Next: filter data using WHERE and GROUP BY.

Remember, use INNER JOIN when you need clean, paired data and LEFT JOIN when you need to preserve your full list. Understanding these two will handle 90 percent of your needs.
