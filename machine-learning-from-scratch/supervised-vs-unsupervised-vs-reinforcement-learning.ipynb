## Supervised Learning Intuition

- Features X map to targets y
- Learns from ground-truth answers
- Goal: minimize prediction error

Supervised learning requires paired data: input features capital X and target labels y. Think of it like a student studying with an answer key at the back of the textbook. The model predicts an answer y hat, compares it against the true label y, and adjusts its internal parameters to minimize the error. If you have ground-truth answers from past historical events, your problem is supervised.

## Supervised Math: Mean Squared Error

$$ \hat{y} = w \cdot x + b $$

$$ L = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2 $$

Here is the core mathematics for regression tasks. In linear regression, the predicted output y hat equals weight w times feature x plus bias b. To measure how poorly our model performs across capital N examples, we calculate the Mean Squared Error, or MSE. The loss L equals one over N times the sum over all i from one to N of the true label y sub i minus predicted y hat sub i, squared.

## Worked MSE Calculation

$$ y = [110, 140, 210], \quad \hat{y} = [100, 150, 200] $$

$$ \mathrm{MSE} = \frac{1}{3} \left( (110 - 100)^2 + (140 - 150)^2 + (210 - 200)^2 \right) $$

$$ \mathrm{MSE} = \frac{1}{3} \left( 10^2 + (-10)^2 + 10^2 \right) $$

$$ \mathrm{MSE} = \frac{1}{3} \left( 100 + 100 + 100 \right) = 100.0 $$

Let us work through a concrete calculation with three factory sites. Actual material consumption y is one hundred ten, one hundred forty, and two hundred ten tons. Our model predicts y hat as one hundred, one hundred fifty, and two hundred tons. The differences are ten, minus ten, and ten. Squaring each gives one hundred, one hundred, and one hundred. Adding them yields three hundred. Dividing by three gives an MSE of exactly one hundred point zero.

## Supervised Regression in Python

Here is the exact Python code to compute this loss using NumPy. We create NumPy arrays for true and predicted values, subtract them elementwise, square the results, and take the mean. Printing MSE produces one hundred point zero, and taking the square root gives a Root Mean Squared Error of ten point zero tons. The Python output matches our hand calculations exactly.

```python
import numpy as np

# Actual vs predicted material usage (tons)
y_true = np.array([110.0, 140.0, 210.0])
y_pred = np.array([100.0, 150.0, 200.0])

# Calculate Mean Squared Error
mse = np.mean((y_true - y_pred) ** 2)
rmse = np.sqrt(mse)

print("MSE:", mse)    # -> MSE: 100.0
print("RMSE:", rmse)  # -> RMSE: 10.0
```

Notice how the trend line passes through our scatter plot of historical data points. Checkpoint one: we have established supervised learning, derived the Mean Squared Error, and validated the calculation with Python code. Next, what happens when you do not have ground-truth answers in your data?

## Unsupervised Learning Intuition

- Unlabeled data matrix X
- Discovers intrinsic data structure
- Groups items by feature similarity

That brings us to unsupervised learning. When my team faced fifty-seven thousand messy material codes during a corporate acquisition, nobody had time or labels to tag duplicates. Unsupervised algorithms take an unlabeled feature matrix capital X and group items by natural geometry and distance in feature space.

## Unsupervised Math: Euclidean Distance

$$ d(p, q) = \sqrt{\sum_{i=1}^{n} (p_i - q_i)^2} $$

$$ d(p, q) = \sqrt{(p_1 - q_1)^2 + (p_2 - q_2)^2} $$

To group unlabeled points into clusters, models measure distance between feature vectors p and q. The standard metric is Euclidean distance. Distance d between vectors p and q equals the square root of the sum over all n features of p sub i minus q sub i squared. For two features, it simplifies to the straight-line Pythagorean formula.

## Worked Distance Calculation

$$ p = [10.0, 50.0], \quad q = [12.0, 48.0] $$

$$ d(p, q) = \sqrt{(10.0 - 12.0)^2 + (50.0 - 48.0)^2} $$

$$ d(p, q) = \sqrt{(-2.0)^2 + (2.0)^2} $$

$$ d(p, q) = \sqrt{4.0 + 4.0} = \sqrt{8.0} \approx 2.83 $$

Let us compute distance between two material code entries. Item p has text length ten characters and weight fifty kilograms. Item q has length twelve characters and weight forty-eight kilograms. Ten minus twelve squared is four. Fifty minus forty-eight squared is four. The sum is eight. The square root of eight equals approximately two point eight three.

## K-Means Clustering in Python

Here is how to group these material features in Scikit-Learn. We define five data items, fit a K-Means model with two clusters, and inspect the resulting cluster assignments. The output is cluster one for the first three similar items and cluster zero for the remaining two items, cleanly isolating duplicate catalog groups without manual tagging.

```python
import numpy as np
from sklearn.cluster import KMeans

# Feature vectors: [name length, weight kg]
X = np.array([[10, 50], [12, 48], [11, 51], [40, 5], [42, 6]])

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
kmeans.fit(X)

print("Cluster Labels:", kmeans.labels_)
# -> Cluster Labels: [1 1 1 0 0]
```

## K-Means Convergence Loop

1. Initialize K Centroids
2. Assign Points to Nearest
3. Recompute Centroid Means
4. Check Convergence

*Iterative centroid optimization without ground truth*

Under the hood, K-Means follows a four-step iterative loop. It initializes centroid locations, assigns each unlabeled vector to the nearest centroid using Euclidean distance, recalculates centroid means, and repeats until cluster centers stop moving.

## Reinforcement Learning Intuition

- Agent interacts with Environment
- Takes Action a from State s
- Receives Reward r and Next State s'

This brings us to Reinforcement Learning. Unlike supervised or unsupervised paradigms that process static datasets, an RL agent learns dynamically inside an environment. Think of training a game agent or an automated inventory replenishment controller. The agent observes state s, takes action a, receives numerical reward r, and transitions to new state s prime.

## RL Math: Q-Learning Update Rule

$$ Q(s, a) \leftarrow Q(s, a) + \alpha \left[ r + \gamma \max_{a'} Q(s', a') - Q(s, a) \right] $$

The foundational algorithm for model-free reinforcement learning is Q-Learning, based on the Bellman Equation. New Q-value for state s and action a equals current Q-value plus learning rate alpha times the temporal difference error. The error is reward r plus discount factor gamma times maximum future Q-value at state s prime, minus current Q-value.

## Worked Q-Value Update Step

$$ Q(s,a)=0.0, \quad \alpha=0.1, \quad r=10.0, \quad \gamma=0.9, \quad \max Q'=5.0 $$

$$ Q(s, a) \leftarrow 0.0 + 0.1 \left[ 10.0 + 0.9 \cdot 5.0 - 0.0 \right] $$

$$ Q(s, a) \leftarrow 0.1 \left[ 10.0 + 4.5 \right] $$

$$ Q(s, a) \leftarrow 0.1 \cdot 14.5 = 1.45 $$

Let us execute one numerical update step. Initial Q-value is zero point zero, learning rate alpha is zero point one, reward r is ten point zero, discount gamma is zero point nine, and maximum expected future reward is five point zero. Discounted future reward is four point five. Total target is fourteen point five. Multiplying by learning rate zero point one yields an updated Q-value of exactly one point four five.

## Q-Table Update in Python

Here is the exact Python implementation for updating a Q-table. We define the table values, calculate maximum future Q-value across next state actions, scale the temporal difference error by alpha, and update the state action pair. The printed Q-value output is one point four five, matching our manual mathematics step by step.

```python
import numpy as np

# Q-table initialization
q_table = np.array([[0.0, 0.0], [5.0, 2.0]])
state, action = 0, 1
reward, alpha, gamma = 10.0, 0.1, 0.9

# Bellman update
best_next = np.max(q_table[1])  # 5.0
q_table[state, action] += alpha * (
    reward + gamma * best_next - q_table[state, action]
)

print("Updated Q-value:", q_table[state, action])
# -> Updated Q-value: 1.45
```

## Agent-Environment Loop

1. State s Observed
2. Action a Selected
3. Environment Transition
4. Reward r Received

*Continuous feedback loop optimizes policy over time*

## Paradigm Selection Matrix

**Which ML Type Fits Your Problem?** (highlighted rows in **bold**)

| Question | Supervised | Unsupervised | Reinforcement |
|---|---|---|---|
| **Have ground truth labels?** | **Yes** | **No** | **No (Rewards only)** |
| Data format | Pairs (X, y) | Features X only | Environment feedback |
| Primary goal | Predict labels | Find structure | Learn optimal actions |
| Primary metric | MSE / Accuracy | Silhouette / Distance | Cumulative return |

Use this decision framework when starting any project. The golden rule is simple: do you have labeled ground-truth answers from historical data? If yes, it is Supervised learning. If you have unlabeled features and need hidden grouping, it is Unsupervised. If your model learns by taking trial actions in a dynamic environment to maximize feedback, it is Reinforcement learning.

## Key Takeaways

- Supervised: maps features X to labels y with loss L
- Unsupervised: finds patterns using feature geometry
- RL: maximizes reward through environment trial
- Golden rule: check for past ground-truth labels

To recap: supervised learning optimizes predictions against ground-truth labels, unsupervised learning discovers structure in unlabeled feature space, and reinforcement learning optimizes action policies via environmental reward signals.
