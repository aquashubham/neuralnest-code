## Measuring Standard Accuracy

$$ \mathrm{Accuracy} = \frac{\mathrm{TP} + \mathrm{TN}}{\mathrm{TP} + \mathrm{TN} + \mathrm{FP} + \mathrm{FN}} $$

*Formula definition*

$$ \mathrm{Accuracy} = \frac{6 + 88}{6 + 88 + 2 + 4} $$

*Value substitution*

$$ \mathrm{Accuracy} = \frac{94}{100} = 0.94 $$

*Final accuracy result*

Let us start with standard accuracy. Accuracy equals true positives plus true negatives, divided by total samples. Substituting our values gives six plus eighty-eight, divided by one hundred. That equals zero point nine four, or ninety-four percent accuracy. On paper, ninety-four percent sounds excellent.

## The Accuracy Paradox

- High accuracy masks critical errors
- Imbalanced datasets distort performance
- LLM screening rejected candidate pool

Ninety-four percent sounds great, but look closer. The model missed four out of ten breakdowns. In production, a single missed breakdown costs fifty thousand dollars in lost production. When I deployed an LLM resume screening tool across six thousand job requisitions, ninety percent of applicants were unqualified. A dumb model that rejects every single resume gets ninety percent accuracy automatically, while hiring zero qualified engineers. Accuracy alone fails on imbalanced data.

## Precision: Quality of Positive Predictions

$$ \mathrm{Precision} = \frac{\mathrm{TP}}{\mathrm{TP} + \mathrm{FP}} $$

*Formula definition*

$$ \mathrm{Precision} = \frac{6}{6 + 2} $$

*Value substitution*

$$ \mathrm{Precision} = \frac{6}{8} = 0.75 $$

*Final precision result*

To fix this, we split evaluation into specific questions. The first metric is Precision. Precision answers: of all the times the model predicted a breakdown, how many were actually breakdowns? Precision equals true positives divided by true positives plus false positives. In our numbers, that is six divided by six plus two. That gives six over eight, which equals zero point seven five, or seventy-five percent.

## Precision Focus Area

1. All Predictions
2. Predicted Positive (8)
3. True Breakdowns (6)

*Precision examines predicted positive column*

Visually, Precision focuses only on the alarms raised by the model. It completely ignores true negatives. If your maintenance team spends three hours inspecting the machine every time an alarm sounds, high precision prevents wasted labor.

## Recall: Completeness of Catching Positives

$$ \mathrm{Recall} = \frac{\mathrm{TP}}{\mathrm{TP} + \mathrm{FN}} $$

*Formula definition*

$$ \mathrm{Recall} = \frac{6}{6 + 4} $$

*Value substitution*

$$ \mathrm{Recall} = \frac{6}{10} = 0.60 $$

*Final recall result*

The second metric is Recall, also called sensitivity. Recall answers: of all the actual breakdowns that occurred, how many did our model catch? Recall equals true positives divided by true positives plus false negatives. Here, six divided by six plus four equals six over ten, or zero point six zero. We caught sixty percent of actual breakdowns.

## Recall Focus Area

1. All Actual Events
2. Actual Breakdowns (10)
3. Caught Breakdowns (6)

*Recall examines actual positive row*

Notice the difference. Precision focuses on model predictions, while Recall focuses on actual events in the ground truth data. High Recall means very few positive cases slipped through undetected.

## Computing Precision and Recall in Python

Here is how you compute Precision and Recall using scikit-learn. We define our true labels array with ten ones and ninety zeros. Then we create predicted labels matching our confusion matrix. Calling precision score yields zero point seven five, and recall score yields zero point six zero. The code output matches our manual calculation exactly.

```python
from sklearn.metrics import precision_score, recall_score

y_true = [1]*10 + [0]*90
# 6 TP, 4 FN, 2 FP, 88 TN
y_pred = [1]*6 + [0]*4 + [1]*2 + [0]*88

prec = precision_score(y_true, y_pred)
rec = recall_score(y_true, y_pred)

print(f"Precision: {prec:.2f}")  # -> Precision: 0.75
print(f"Recall: {rec:.2f}")        # -> Recall: 0.60
```

## Precision-Recall Trade-off

Can we get one hundred percent Precision and one hundred percent Recall at the same time? Almost never. Look at this threshold curve. Binary classifiers output a probability between zero and one. If we lower the threshold to zero point one, we flag everything as a breakdown. Recall jumps to one point zero, but Precision plummets to zero point two five because of false alarms.

```python
import matplotlib.pyplot as plt

x = [0.1, 0.3, 0.5, 0.7, 0.9]
plt.plot(x, [0.25, 0.5, 0.75, 0.88, 1.0], marker='o', label='Precision')
plt.plot(x, [1.0, 0.9, 0.6, 0.3, 0.1], marker='o', label='Recall')
plt.xlabel('Decision Threshold')
plt.ylabel('Score')
plt.legend()
plt.show()
```

## Adjusting the Decision Threshold

In Python, classification models return probabilities using predict proba. By default, scikit-learn cuts off at zero point five. If we lower our threshold to zero point two, our condition tags more items as positive. Recall leaps from sixty percent to ninety percent, catching almost every breakdown. But Precision falls to thirty-eight percent.

```python
import numpy as np
from sklearn.metrics import precision_score, recall_score

# Simulated model probability output for 100 samples
probs = np.array([0.9, 0.85, 0.8, 0.7, 0.65, 0.55, 0.4, 0.35, 0.3, 0.25] + [0.05]*90)

# Lower threshold to 0.2 to catch more breakdowns
custom_pred = (probs >= 0.2).astype(int)

print("Recall:", recall_score(y_true, custom_pred))      # -> Recall: 0.90
print("Precision:", precision_score(y_true, custom_pred)) # -> Precision: 0.38
```

## Choosing Based on Business Cost

- High Recall: Medical diagnosis and factory failures
- High Precision: Spam filtering and loan approvals
- Match metric to misclassification penalty

So how do you choose between Precision and Recall? You look at the cost of mistakes. In medical screening or machine breakdown, a false negative kills people or breaks factories. You optimize for high Recall. In spam detection, sending an important job offer to spam is far worse than seeing one extra promotion in your inbox. There, you optimize for high Precision.

## The F1 Score Definition

$$ F_1 = \frac{2}{\frac{1}{\mathrm{Precision}} + \frac{1}{\mathrm{Recall}}} $$

*Harmonic mean definition*

$$ F_1 = \frac{2}{\frac{\mathrm{Recall} + \mathrm{Precision}}{\mathrm{Precision} \cdot \mathrm{Recall}}} $$

*Algebraic expansion*

$$ F_1 = 2 \cdot \frac{\mathrm{Precision} \cdot \mathrm{Recall}}{\mathrm{Precision} + \mathrm{Recall}} $$

*Simplified formula*

That single metric is the F1 Score. The F1 Score is the harmonic mean of Precision and Recall. Instead of a standard average, the harmonic mean heavily punishes extreme values. The top formula shows two divided by one over Precision plus one over Recall. Rearranging terms gives two times Precision times Recall, divided by Precision plus Recall.

## Calculating the F1 Score

$$ F_1 = 2 \cdot \frac{0.75 \cdot 0.60}{0.75 + 0.60} $$

*Substitute Precision and Recall*

$$ F_1 = 2 \cdot \frac{0.45}{1.35} $$

*Multiply terms*

$$ F_1 = \frac{0.90}{1.35} = 0.667 $$

*Final F1 result*

Let us compute the F1 Score for our factory model. We substitute zero point seven five for Precision and zero point six zero for Recall. Zero point seven five times zero point six zero equals zero point four five. Two times zero point four five is zero point nine zero. Dividing zero point nine zero by one point three five gives zero point six six seven.

## Harmonic Mean vs Simple Average

Why do we use the harmonic mean instead of a simple arithmetic mean? Suppose a model has a Precision of one point zero but a Recall of zero. A simple average gives fifty percent score, hiding total model failure. The harmonic mean gives an F1 Score of zero. If either metric collapses, the F1 Score collapses too.

```python
import matplotlib.pyplot as plt

labels = ['Precision 1.0, Recall 0.0', 'Precision 0.75, Recall 0.60']
width = 0.8 / 2
plt.bar([i - 0.4 + width * (0 + 0.5) for i in range(2)], [0.5, 0.675], width, label='Arithmetic Mean')
plt.bar([i - 0.4 + width * (1 + 0.5) for i in range(2)], [0.0, 0.667], width, label='F1 Score (Harmonic)')
plt.xticks(range(2), labels)
plt.xlabel('Model Scenario')
plt.ylabel('Metric Score')
plt.legend()
plt.show()
```

## Scikit-Learn Classification Report

In professional workflow, you do not compute these metrics individually. You run scikit-learn classification report. Passing y true and y pred outputs Precision, Recall, and F1 Score for every class, alongside macro and weighted averages. This gives an immediate full health check of your model.

```python
from sklearn.metrics import classification_report

y_true = [1]*10 + [0]*90
y_pred = [1]*6 + [0]*4 + [1]*2 + [0]*88

# Generate full evaluation summary
report = classification_report(y_true, y_pred, digits=3)
print(report)
# Output shows precision (0.750), recall (0.600), f1-score (0.667)
```

## The Common Evaluation Mistake

- Optimizing F1 blindly without cost analysis
- Failing to report full classification metrics
- Ignoring threshold tuning in production

The most common mistake junior developers make is optimizing F1 score blindly without consulting domain experts. In real projects, always calculate the financial cost of false positives versus false negatives. Pick your metric based on cost, and report all four metrics together.
