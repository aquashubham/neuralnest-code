## Hardcoded Rules Fail

Writing fifty hardcoded if-else statements to predict machine failure breaks the second real production data hits.

```python
def fails(h, t):
    return h > 3500 and t > 90

print(fails(3600, 95))  # True
print(fails(3400, 99))  # False!
```
