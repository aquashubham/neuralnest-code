This Python function passes code review, then quietly remembers data between calls. Here's why mutable default arguments are created only once, and the one-line fix.

## Why this matters

This Python function has a bug that passes most code reviews. Call it twice, and it remembers data it should have forgotten. Watch.

## The bug

add takes a number and a list that defaults to empty. The first call returns one, as expected. The second call should return just two. Instead, you get one and two. Python creates that default list once, when the function is defined, and every call reuses the same list.

```python
def add(x, items=[]):
    items.append(x)
    return items

print(add(1))  # -> [1]
print(add(2))  # -> [1, 2]
```

## The fix

- Default to None: items=None
- Inside, create a new empty list
- Every call gets a fresh list

The fix: default to None. Inside the function, if items is None, create a new empty list. Now every call gets its own fresh list, and add of two returns just two.

## Remember

- Defaults are created once
- Never default to a list or dict

Rule of thumb: never use a list or a dictionary as a default argument. Defaults are created once, not on every call. Follow for more Python traps.
