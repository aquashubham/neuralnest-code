## The Function That Remembers

This function remembers things it shouldn't. Second call prints a, b. Why? Python builds the default list once, at definition. Every call shares it.

```python
def add(item, cart=[]):
    cart.append(item)
    return cart

print(add('a'))  # -> ['a']
print(add('b'))  # -> ['a', 'b']
```

## The Fix

Fix: default to None, create the list inside. Defaults run once. Not every call.

```python
def add(item, cart=None):
    if cart is None:
        cart = []
    cart.append(item)
    return cart
```
