## List vs Set Lookup Speed

This Python lookup runs fifty thousand times slower than needed. Checking a list scans every single element one by one.

```python
codes = list(range(57000))
# Slow list check: scans all
56999 in codes  # -> slow

codes_set = set(codes)
# Fast set check: instant
56999 in codes_set  # -> fast
```

## Use Sets for Instant Search

- Lists scan items linearly
- Sets use hashing for instant lookups

Convert your list to a set first. Set lookups take one step, making searches instant.
