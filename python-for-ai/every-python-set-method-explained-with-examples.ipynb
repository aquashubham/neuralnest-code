## Creating, Copying, and Clearing Sets

Let us start with set creation and a classic trap. Empty curly braces do not create an empty set; they create an empty dictionary. To create an empty set, you must call the set function with parentheses. Next, when you call copy on a set, Python creates a shallow duplicate in memory. If you clear the original set using clear, the original set becomes empty, shown as set in Python print output, but your copied set retains all original values intact. Notice that clear modifies the set in place and returns None.

```python
# Empty set gotcha
empty_dict = {}
empty_set = set()

codes_a = {"M-101", "M-102", "M-103"}
backup_codes = codes_a.copy()
codes_a.clear()

print(type(empty_dict))  # -> <class 'dict'>
print(codes_a)           # -> set()
print(backup_codes)      # -> {'M-101', 'M-102', 'M-103'}
```

## Set Creation and Basics Summary

**Basic Set Operations** (highlighted rows in **bold**)

| Operation | Syntax | Returns | Mutates? |
|---|---|---|---|
| Create Empty Set | set() | new set | No |
| **Empty Dict Trap** | **{}** | **new dict** | **No** |
| Duplicate Set | set_a.copy() | new set copy | No |
| **Clear All Items** | **set_a.clear()** | **None** | **Yes** |

Look at this summary table. When creating an empty set, set creates a new empty set object, while empty curly braces create a dictionary. Method copy returns a distinct new set object without modifying the existing set. On the other hand, method clear removes all elements, mutates the original set in place, and returns None. In Python, methods that modify data structures in place almost always return None, which prevents accidental method chaining bugs in your code.

## Adding Items: add vs update

When adding new material codes to your set, you have two methods: add and update. Method add takes exactly one item and inserts it into the set. If that item is already present, like M-101 in our snippet, Python simply ignores it without raising an error. If you have multiple items to add, such as a list of new material numbers from a spreadsheet, pass that list to update. Method update accepts any iterable like lists, tuples, or sets, adding every unique element. Both add and update modify the set in place and return None.

```python
codes = {"M-101", "M-102"}

# Add single item
codes.add("M-104")
# Adding duplicate does nothing
codes.add("M-101")

# Update with multiple items (iterable)
codes.update(["M-105", "M-106"])

print(codes)
# -> {'M-101', 'M-102', 'M-104', 'M-105', 'M-106'}
```

## Removing Items: remove, discard, pop

Removing items is where developers often hit runtime crashes. You have three options: discard, pop, and remove. Method discard attempts to remove an item, but if the item does not exist, like M-999, it does nothing and continues quietly. Method remove also attempts to remove an item, but if that item is missing, it raises a KeyError. Use remove when a missing item represents corrupted data that should stop execution. Finally, pop removes and returns one single arbitrary item. Because sets are unordered, you cannot predict which item pop will pick.

```python
codes = {"M-101", "M-102", "M-103"}

# discard: safe removal
codes.discard("M-999")  # No error!

# pop: removes arbitrary item
popped = codes.pop()

# remove: raises error if missing
try:
    codes.remove("M-999")
except KeyError:
    print("KeyError caught!")  # -> KeyError caught!
```

## Combining Sets: New Set Operations

To combine data between Company A and Company B without altering original sets, use non-mutating set operations. Method union or the pipe operator combines all unique items from both sets. Method intersection or the ampersand operator returns only items present in both sets, which is M-103 here. Method difference or the minus operator returns items in set A that are not in set B. Finally, symmetric_difference or the caret operator returns items present in either set, but not in both. All four methods return a brand new set object.

```python
a = {"M-101", "M-102", "M-103"}
b = {"M-103", "M-104", "M-105"}

all_codes = a.union(b)                 # or a | b
shared = a.intersection(b)             # or a & b
only_a = a.difference(b)               # or a - b
unique_to_each = a.symmetric_difference(b) # or a ^ b

print(all_codes)       # -> {'M-101', 'M-102', 'M-103', 'M-104', 'M-105'}
print(shared)          # -> {'M-103'}
print(only_a)          # -> {'M-101', 'M-102'}
print(unique_to_each)  # -> {'M-101', 'M-102', 'M-104', 'M-105'}
```

## Modifying Sets In Place

We are now halfway through our set methods guide, moving from creating new sets to in-place modifications. Creating new sets consumes memory when processing millions of records. That is where in-place update methods help. Method intersection_update, or ampersand equals, modifies set A so it only retains elements also found in set B. Method difference_update, or minus equals, removes all elements from set A that exist in set B. There is also symmetric_difference_update, or caret equals. Notice that all update methods modify targeted sets in place and return None.

```python
a = {"M-101", "M-102", "M-103"}
b = {"M-103", "M-104"}

# Retain only common elements in 'a'
a.intersection_update(b)  # or a &= b
print(a)  # -> {'M-103'}

c = {"M-101", "M-102", "M-103"}
# Remove elements found in 'b'
c.difference_update(b)     # or c -= b
print(c)  # -> {'M-101', 'M-102'}
```

## Comparing Sets: Subsets and Disjoint

To evaluate relationships between datasets, Python provides three boolean comparison methods: issubset, issuperset, and isdisjoint. Method issubset, or less-than-or-equal operator, returns True if every element in the first set exists in the second set. Method issuperset, or greater-than-or-equal operator, checks the exact reverse. Method isdisjoint returns True if two sets have zero elements in common. For example, core codes and other codes share no material numbers, so core dot isdisjoint other evaluates to True. These boolean checks return True or False without modifying data.

```python
core = {"M-101", "M-102"}
full = {"M-101", "M-102", "M-103"}
other = {"M-201", "M-202"}

print(core.issubset(full))     # -> True (or core <= full)
print(full.issuperset(core))   # -> True (or full >= core)
print(core.isdisjoint(other))  # -> True (no overlap)
```

## Top Set Traps to Avoid

- Using {} creates dict instead of set
- pop removes arbitrary item not last
- remove throws KeyError on missing items
- In-place update methods return None

Let us review the four most common mistakes developers make with Python sets. First, writing empty curly braces creates an empty dictionary, not an empty set. Always use set parentheses for empty sets. Second, calling pop on a set does not remove the last item because sets are unordered; it removes an arbitrary item. Third, using remove on an item that does not exist crashes your program with a KeyError; use discard if missing items are expected. Fourth, assigning the result of an update method to a variable sets it to None.

This cheat sheet summarizes key set operations you will use daily in Python data engineering and AI preprocessing. They all return a brand new set object without mutating original sets. In contrast, item modification methods like add, remove, discard, and update mutate the target set directly and return None.
