## Cleaning Messy Strings

Let us start with string manipulation using our material code example. Here we have three different raw strings: uppercase with whitespace, comma separated with dashes, and lowercase with extra trailing spaces. They all represent the exact same hardware component. We define a clean code function that takes a string input. We apply strip to remove outside space, lower to unify case, and chained replace methods to strip out commas and dashes. When we call clean code on the second item, it returns the unified string bolt m12. Standardizing strings early prevents duplicated records from reaching your language model.

```python
raw_codes = ["  BOLT M12", "Bolt, M-12", "bolt m12 "]

def clean_code(text: str) -> str:
    cleaned = text.strip().lower()
    cleaned = cleaned.replace(",", "").replace("-", "")
    return cleaned

print(clean_code(raw_codes[1]))
# -> bolt m12
```

## List Comprehensions

Now that we have a function to clean one string, how do we apply it to thousands of items efficiently? Python list comprehensions provide a clean syntax for transforming collections. The expression inside the square brackets runs clean code for every element in raw codes and builds a new list in memory. The printed output shows three identical clean strings. Passing that resulting list into Python set function automatically drops all duplicate elements, leaving a single unique entry. This pattern replaces long traditional for loops with one expressive line.

```python
def clean_code(code):
    code = code.strip().lower()
    code = code.replace(",", "").replace("-", "")
    return " ".join(code.split())

raw_codes = ["  BOLT M12", "Bolt, M-12", "bolt m12 "]

cleaned = [clean_code(code) for code in raw_codes]
print(cleaned)
# -> ['bolt m12', 'bolt m12', 'bolt m12']

unique_codes = set(cleaned)
print(unique_codes)
# -> {'bolt m12'}
```

## Parsing JSON API Data

Most web interfaces and language model integrations communicate using JSON strings. Python built in json module converts these strings into native dictionaries using json loads. Here, raw json contains a serialized string with stock details. Passing it to json loads yields a real Python dictionary. We access the stock keeping unit key directly to get bolt m12. The type output confirms Python converted the payload into a native dictionary. But what happens if the incoming JSON missing a field or contains an unexpected data type? Let us look at structured validation next.

```python
import json

raw_json = '{"sku": "bolt m12", "price": 1.5, "stock": 500}'
data = json.loads(raw_json)

print(data["sku"])
# -> bolt m12

print(type(data))
# -> <class 'dict'>
```

## Data Validation with Pydantic

Pydantic has become the standard data validation library in modern Python data engineering and framework tools like FastAPI. We define a class named Product that inherits from Pydantic BaseModel. We annotate sku as a string, price as a floating point number, and stock as an integer. When we instantiate Product, Pydantic checks every argument against those types automatically. Accessing item dot price gives typed attributes directly. Calling model dump converts the validated object back into a clean standard dictionary.

```python
from pydantic import BaseModel

class Product(BaseModel):
    sku: str
    price: float
    stock: int

item = Product(sku="bolt m12", price=1.5, stock=500)
print(item.price)
# -> 1.5
print(item.model_dump())
# -> {'sku': 'bolt m12', 'price': 1.5, 'stock': 500}
```

## Sequential vs Async API Calls

Once your data models are validated, you will often need to send hundreds of prompts or requests to remote endpoints. If you send requests sequentially, your code pauses and waits for network responses one by one. The blue bars on this chart show execution time scaling linearly to fifty seconds for one hundred requests. The green bars show the exact same requests executed asynchronously using Python asyncio module. Time remains under one second because the program sends new requests while waiting for network returns. Let us look at the code for concurrent requests.

```python
import matplotlib.pyplot as plt

labels = ['10 Requests', '50 Requests', '100 Requests']
width = 0.8 / 2
plt.bar([i - 0.4 + width * (0 + 0.5) for i in range(3)], [5.0, 25.0, 50.0], width, label='Sequential')
plt.bar([i - 0.4 + width * (1 + 0.5) for i in range(3)], [0.6, 0.7, 0.8], width, label='AsyncIO Parallel')
plt.xticks(range(3), labels)
plt.xlabel('Batch Size')
plt.ylabel('Execution Time (seconds)')
plt.legend()
plt.show()
```

## Async API Requests with Asyncio

Asyncio allows Python to pause execution on slow tasks like network requests without blocking the entire program. We mark our fetch function with the async keyword and use await on asynchronous sleep to simulate network wait times. Inside main, asyncio gather schedules fetch product for item one and item two concurrently. Python starts the second request immediately instead of idling on the first. The program gathers all finished values and prints the list of outputs. Master this basic pattern to scale up network requests.

```python
import asyncio

async def fetch_product(item_id: int):
    await asyncio.sleep(0.01)
    return f"item_{item_id}"

async def main():
    results = await asyncio.gather(
        fetch_product(1), fetch_product(2)
    )
    print(results)
    # -> ['item_1', 'item_2']

asyncio.run(main())
```

## Loading Data into Pandas

Once you have cleaned and validated your dictionary structures, converting them into Pandas DataFrames takes a single line. Here, we import pandas as pd and pass our list of dictionaries straight into DataFrame. Shape returns two rows and three columns, matching our input dictionaries. We access the price column directly and compute the average price using mean, returning one point one five. Clean Python dictionaries transition smoothly into tabular data objects.

```python
import pandas as pd

data = [
    {"sku": "bolt m12", "price": 1.5, "stock": 500},
    {"sku": "nut m10", "price": 0.8, "stock": 1200}
]

df = pd.DataFrame(data)
print(df.shape)
# -> (2, 3)
print(df["price"].mean())
# -> 1.15
```
