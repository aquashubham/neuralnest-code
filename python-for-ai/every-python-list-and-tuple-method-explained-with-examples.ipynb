## Lists vs Tuples: Mutability

- Lists are dynamic arrays that can change
- Tuples are fixed sequences that cannot change
- Lists offer eleven methods for data cleanup
- Tuples offer only two searching methods

This video is the third lesson in our Python for AI series, building the bridge between core Python programming and data libraries like NumPy and Pandas. Before touching big data tools, you must master core data structures. Lists are dynamic arrays. You can add items, remove items, reorder items, and replace items. Because lists are mutable, Python provides eleven built-in methods to manipulate them. Tuples are immutable. Once created, their size and contents are locked forever in memory. That is why tuples have only two methods, while lists have eleven. Let us start by adding new material codes to a list.

## Adding Elements to Lists

We start with our material codes list containing two items. The first adding method is append. Calling append with MAT 103 places that single element at the very end of the list. Notice that append returns None and mutates the list in place. Next is extend. When we have a second list of material codes named new batch, calling extend adds each individual item from new batch into codes. If we had used append here instead, we would have added a single nested list inside our list. Finally, insert takes an index and an item. Calling insert with zero and MAT 100 pushes MAT 100 to the front.

```python
codes = ["MAT-101", "MAT-102"]
codes.append("MAT-103")
# -> ['MAT-101', 'MAT-102', 'MAT-103']

new_batch = ["MAT-104", "MAT-105"]
codes.extend(new_batch)
# -> ['MAT-101', 'MAT-102', 'MAT-103', 'MAT-104', 'MAT-105']

codes.insert(0, "MAT-100")
# -> ['MAT-100', 'MAT-101', 'MAT-102',
#     'MAT-103', 'MAT-104', 'MAT-105']
```

## Summary: Adding Methods

**List Insertion Methods** (highlighted rows in **bold**)

| Method | Return Value | Mutates In-Place | Behavior |
|---|---|---|---|
| **append(x)** | **None** | **Yes** | **Adds single item to end** |
| **extend(iterable)** | **None** | **Yes** | **Appends each item from iterable** |
| **insert(i, x)** | **None** | **Yes** | **Inserts item at index i** |

Let us compare these three insertion methods side by side. Notice the highlighted rows in our table. All three methods return None while directly altering the original list in memory. The key difference lies in what gets added. Append takes one item and puts it at the end. Extend takes an iterable, such as another list or tuple, and unpacks every element onto the end. Insert lets you place an item at any specific position, though inserting at index zero forces Python to shift every existing element to the right. Next, what happens when we need to remove bad material codes from our merged dataset?

## Removing Elements from Lists

To clean up bad data, Python gives us three removal methods: remove, pop, and clear. Look at our list containing corrupt entries named BAD 999. Calling remove with BAD 999 searches from left to right and deletes the very first matching element it finds. It returns None. Notice that the second BAD 999 is still in the list. Next, pop takes an optional index position. Calling pop with index two removes the element at position two and returns that removed value. If you call pop without an argument, it removes and returns the last item. Lastly, clear empties the list completely, returning None and leaving an empty list.

```python
codes = ["MAT-100", "BAD-999", "MAT-101", "BAD-999"]
codes.remove("BAD-999")
# -> ['MAT-100', 'MAT-101', 'BAD-999']

removed_item = codes.pop(2)
# -> 'BAD-999'
# codes is now ['MAT-100', 'MAT-101']

codes.clear()
# -> []
```

## Summary: Removing Methods

**List Deletion Methods** (highlighted rows in **bold**)

| Method | Return Value | Mutates In-Place | Error Trigger |
|---|---|---|---|
| **remove(x)** | **None** | **Yes** | **ValueError if x not found** |
| **pop(i)** | **Removed Item** | **Yes** | **IndexError if out of range** |
| clear() | None | Yes | None |

Let us examine the table of deletion methods. Look at the highlighted error triggers. Remove searches for a value and deletes its first match, but if that value is missing, Python raises a ValueError and halts execution. Pop removes by index position and returns the item, but if the index is out of bounds or the list is empty, Python raises an IndexError. Clear simply deletes every element without raising errors. So remove searches by value and returns None, pop selects by position and returns the item, and clear empties everything. Now, what if we want to locate codes without removing them?

## Searching Lists: index and count

When searching a dataset, index and count are your core tools. Neither method changes the list. They both return numbers. The index method takes a value and returns the zero-based index of its first occurrence. Here, MAT 101 is at index one. You can also pass start and end positions to limit where index searches. Calling index with MAT 100 starting from position one returns index two, skipping position zero. The count method counts how many times a value appears in the entire list. Here, MAT 100 appears twice, so count returns two. Just like remove, if index cannot find the target value, it raises a ValueError.

```python
codes = ["MAT-100", "MAT-101", "MAT-100", "MAT-102"]
idx = codes.index("MAT-101")
# idx -> 1

cnt = codes.count("MAT-100")
# cnt -> 2

# Searching range: index(value, start, stop)
idx_part = codes.index("MAT-100", 1, 4)
# idx_part -> 2
```

## Reordering Lists: sort and reverse

To organize our material codes, Python provides sort and reverse. Calling codes dot sort modifies the original list directly, arranging strings alphabetically or numbers in ascending order. It returns None. You can pass reverse equals True to sort in descending order. The reverse method is different. It does not look at the values or sort anything. It simply flips the current order of the list in place, end to end, and returns None. Both sort and reverse mutate the existing list in place. A huge trap here is assigning the result of sort to a variable, which sets your variable to None.

```python
codes = ["MAT-103", "MAT-101", "MAT-102"]
codes.sort()
# -> ['MAT-101', 'MAT-102', 'MAT-103']

codes.sort(reverse=True)
# -> ['MAT-103', 'MAT-102', 'MAT-101']

codes.reverse()
# -> ['MAT-101', 'MAT-102', 'MAT-103']
```

## Copying Lists: Shallow Copy

The final list method is copy. Calling original dot copy creates a shallow copy of the list. This means Python creates a new list container for the outer items. Changing shallow index zero to MAT 999 does not affect original. However, look at index one, which is an inner nested list of warehouse locations. Because copy is shallow, both lists still point to the exact same inner list in memory. Adding LOC C to shallow index one also modifies original index one. If you need nested lists to be completely independent, you must import the copy module and use copy dot deepcopy instead of the built-in copy method.

```python
import copy

original = ["MAT-100", ["LOC-A", "LOC-B"]]
shallow = original.copy()

shallow[0] = "MAT-999"
shallow[1].append("LOC-C")

# original -> ['MAT-100', ['LOC-A', 'LOC-B', 'LOC-C']]
# shallow  -> ['MAT-999', ['LOC-A', 'LOC-B', 'LOC-C']]
```

## Shallow vs Deep Copy Structure

1. Original List
2. Outer Copy Created
3. Shared Inner References
4. Deep Copy Allocates New

Let us trace how memory allocation works during a copy operation. Follow step one to step four in our diagram. Step one shows our original list referencing both string items and an inner nested list. Step two shows what happens when we run the copy method: a new outer list object is created. But step three highlights the pitfall: the outer copy still points directly to the memory addresses of the inner objects. Step four shows that only deep copy recursively duplicates every nested structure. Right around here in our data pipeline, we often convert cleaned material lists into immutable tuples for safety.

## Tuple Methods: count and index

Why do tuples have only two methods? Because tuples are immutable. Once created in memory, their size and elements can never be changed, added, or removed. Therefore, methods like append, pop, sort, or clear do not exist on tuples. If you try to call append on a tuple, Python raises an AttributeError. The only methods tuples have are count and index. Count returns how many times a value occurs, and index returns the first position of a value. Tuples give you write protection and slightly better performance for fixed sequences, such as database schema fields or fixed material code categories.

```python
# Fixed metadata tuple from merger
merged_tuple = ("MAT-100", "MAT-101", "MAT-100")

cnt = merged_tuple.count("MAT-100")
# cnt -> 2

idx = merged_tuple.index("MAT-101")
# idx -> 1

# Modifying triggers AttributeError:
# merged_tuple.append("MAT-102") -> Error!
```

## Performance: Lookups vs Modifications

Understanding performance helps you write fast data engineering scripts. Look at our chart comparing operation scaling across operations on a list of fifty-seven thousand material codes. Direct index lookup and appending to the end are constant time operations, taking O of 1 steps. Inserting an item at index zero forces Python to shift all fifty-seven thousand items, making it linear time, O of N. Sorting takes O of N log N steps, which grows faster as your dataset expands. Knowing these costs helps you decide whether to append data dynamically or build arrays in bulk before sorting.

```python
import matplotlib.pyplot as plt

labels = ['Index Lookup', 'Append', 'Insert at Start', 'In-Place Sort']
width = 0.8 / 1
plt.bar([i - 0.4 + width * (0 + 0.5) for i in range(4)], [1.0, 1.0, 5.0, 12.0], width, label='Big-O Step Count')
plt.xticks(range(4), labels)
plt.xlabel('Operation Type')
plt.ylabel('Time Complexity (O)')
plt.show()
```

## Avoid These 4 Common List Traps

Now let us look at four classic bugs that ruin data cleanup pipelines. Trap number one: writing codes equals codes dot sort. Because sort modifies in place and returns None, this overwrite destroys your dataset completely! Use sorted with parenthesis if you want a new sorted list returned. Trap number two: passing a list into append creates a nested sublist instead of flattening your elements. Use extend when adding multiple elements. Trap number three: calling remove on a code that is missing causes your script to crash with a ValueError. Check with the in keyword first. And trap four: assuming copy duplicates nested lists deep inside your structure.

```python
# Trap 1: sort() returns None
codes = ["MAT-102", "MAT-101"]
codes = codes.sort() # WRONG! codes is now None!
# Correct: codes.sort() OR codes = sorted(codes)

# Trap 2: append() list vs extend()
items = ["A"]
items.append(["B", "C"]) # -> ['A', ['B', 'C']]
# Correct: items.extend(["B", "C"]) -> ['A', 'B', 'C']

# Trap 3: remove() missing item raises ValueError
# Trap 4: copy() is shallow for nested lists
```

## Master Methods Cheat Sheet

**List and Tuple Built-in Methods** (highlighted rows in **bold**)

| Method | Type | Returns | Mutates In-Place |
|---|---|---|---|
| **append / extend / insert** | **List** | **None** | **Yes** |
| **remove / pop / clear** | **List** | **None / Item / None** | **Yes** |
| **sort / reverse** | **List** | **None** | **Yes** |
| **copy** | **List** | **New List** | **No (Shallow)** |
| **count / index** | **List & Tuple** | **Int** | **No** |

Here is your ultimate cheat sheet summarizing all eleven list methods and two tuple methods. Highlighted on screen are the key categories. Remember the rule: if a method modifies a list in place, such as append, extend, insert, remove, sort, or reverse, it returns None. If a method extracts or computes information, such as pop, count, or index, it returns an item or an integer. Tuples only support count and index because they cannot be mutated.
